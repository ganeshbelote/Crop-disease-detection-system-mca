# Dataset exploration
Class distribution of `ml/data/raw` (or point `DATA_DIR` at `../data/demo_synthetic` for a smoke test).

```python
import sys, os
sys.path.insert(0, '../src')
import matplotlib.pyplot as plt
from dataset import discover_classes, build_file_index, class_distribution, stratified_split
DATA_DIR = '../data/raw'
classes = discover_classes(DATA_DIR)
paths, labels, _ = build_file_index(DATA_DIR, classes)
dist = class_distribution(labels, classes)
plt.figure(figsize=(8,4)); plt.bar(dist.keys(), dist.values()); plt.xticks(rotation=45, ha='right'); plt.tight_layout(); plt.show()
train, val, test = stratified_split(paths, labels)
print(len(train.paths), len(val.paths), len(test.paths))
```
